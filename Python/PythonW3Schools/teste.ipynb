##### **1 - CRIAÇÃO DE VARIÁVEL**

Em python, uma variável é criada quando um valor é atribuído nela.

```python
x = 5 
y = "Hello, World!" 
print (x)
```

##### **2 - INSTRUÇÃO**

Uma instrução em python geralmente termina quando a linha termina, não é necessário usar ponto e vírgula.

```python
print("Python is fun!")
```

As instruções são executadas uma após a outra, na mesma ordem em que foram escritas.

```python
print("Hello World!")
print("Have a good day.")
print("Learning Python is fun!")
```

##### **3 - PRINTAR TEXTO**

Por padrão, a função **print()** termina em uma nova linha, para printar multiplas palavras na mesma linha, use o parâmetro end.

```python
print("Hello World!", end=" ")
print("I will print on the same line.")
```

##### **4 - PRINTAR NÚMEROS**

Para printar números, basta escrevê-los dentro dos parênteses, sem aspas.

```python
print(3)
print(358)
print(50000)
```

Para misturar texto e números, separe-os com vírgula.

```python
print("I am", 24, "year old.")
```

##### **5 - VARIÁVEIS**

Uma variável é criada no momento em que você atribui um valor à ela pela primeira vez.

```python
x = 5
y = "John"
print(x)
print(y)
```

**5.1 - CASTING**

Se você quiser especificar qual tipo uma variável terá, utilize casting.

```python
x = str(3)    # x será '3'
y = int(3)    # y será 3
z = float(3)  # z será 3.0
print(x)
print(y)
print(z)
```

**5.2 - SABER O TIPO DE DADO**

É possível saber o tipo de dado de uma variável utilizando a função type().


```python
x = 5
y = "John"
print(type(x))
print(type(y))
```

É possível atribuir valores para múltiplas variáveis em apenas uma linha.

```python
x, y, z = "Orange", "Banana", "Cherry"
print(x)
print(y)
print(z)
```

Também é possível o mesmo valor para múltiplas variáveis.

```python
x = y = z = "Orange"
print(x)
print(y)
print(z)
```

**5.3 - DESEMPACOTAR UMA COLEÇÃO**

Se tiver uma coleção de valores em uma lista, tupla etc, python permite extrair os valores em variáveis.



```python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits
print(x)
print(y)
print(z)
```

**5.4 - VARIÁVEIS GLOBAIS**

Variáveis podem ser criadas fora de funções, e usadas dentro delas.

```python
x = "awesome"

def myfunc():
    print("Python is " + x )

myfunc()   
```

Também é possível criar dentro de uma função, com o mesmo nome de uma variável global.

```python
x = "awesome"

def myfunc():
    x = "fantastic"
    print("Python is " + x)

myfunc()

print("Python is " + x)
```
