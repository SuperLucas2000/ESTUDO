INSTRUÇÃO

Uma instrução em python geralmente termina quando a linha termina, não é necessário usar ponto e vírgula.

```python
print ("Python is fun!")
```

As instruções são executadas uma após a outra, na mesma ordem em que foram escritas.

```python
print("Hello World!")
print("Have a good day.")
print("Learning Python is fun!")
```

PRINTAR TEXTO

Por padrão, a função print() termina em uma nova linha, para printar multiplas palavras na mesma linha, use o parâmetro end.

```python
print("Hello World!", end=" ")
print("I will print on the same line.")
```
